```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# import 5_channel_v1.csv 
data = pd.read_csv('data_3-9/test_4.csv')

# 6.25 mHz sampling rate, 12 bit resolution, 3.3v reference voltage, 5 channels
time = np.arange(len(data)) * (1/6.25)  # time vector based on sampling rate

data = data * (3.3 / 4095)  # convert to voltage

# crop first 100 samples
# data = data[200:]
# time = time[200:]
```

```python
# plot iteration_0 to interation_5 on 6 subplots
fig, axs = plt.subplots(2, 2, figsize=(15, 10))
for i in range(4):  
    axs[i//2, i%2].plot(time, data[f'iteration_{i+3}'], label=f'iteration_{i}')
    axs[i//2, i%2].set_title(f'Iteration {i}')
    axs[i//2, i%2].set_xlabel('Time (us)')
    axs[i//2, i%2].set_ylabel('Voltage (v)')
    axs[i//2, i%2].legend()
plt.tight_layout()
plt.show()
```
